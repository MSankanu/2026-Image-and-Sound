***Methods***

In class we said Python was object-oriented. ***It is about making and evolving objects.*** We have so far seen types of objects that include numbers and strings, but soon we'll see spreadsheets and PDF documents and social networks.

Software objects are like objects in the real world in that they contain data and functionalilty. Think of the chair you're sitting on. There's data about its color (maybe RGB or Pantone). The seat height might be regulated by OSHA. So a chair is data. It's also functionality. Maybe you can adjust the arm heights or roll it around the room. So objects in the real world have data and function.

**Software objects are the same.**

Objects in Python have both *data* as well as *methods*.** Methods are actions you can take on an object. The data and methods associated with an object are called its *attributes*, and, as we have seen, you access them through so-called *dot notation.* As an example, we can use the method `count()` of a string object to count the number of letters (or other sequences of characters) inside a string.

Here is the number of "a"'s in the variable `comment` (defined in the cell above), the string representing a single public comment about the new census regulations.

```python
comment.count("a")
```

...or the number of times "not" appears in the string.

```python
comment.count("not")
```

See how `count()` references the string you're counting in using the "." dot. Change the "a" or "not" above to some other string and see that the counts are correct.

What other things might we want to do with strings? Python has included methods, for example, that create a new string by changing the case of `comment`. In the first cell below we create a string with all caps, and in the second we force all the letters to lowercase.

```python
comment.upper()
```

```python
comment.lower()
```

The method `count()` returned an integer of the number of times "e", say, is in a string. The methods `upper()` and `lower()` create a new string that is all caps or all lowercase. In both cases the original string is unchanged.

Below we assign the output from `upper()` to a new variable called `cap_comment` because its, well, all caps 😊.

```python
cap_comment = comment.upper()

# have a look at the two
print(comment)
print(cap_comment)
```



Another useful method of a string object creates a new string by making a replacement, swapping one sequence of characters in the string for another. Here we swap out the word "Census" in `comment` to create a new string that uses "Gallup poll" instead.

Notice that this method returns a *new* string that has the replacement made. The original string in `comment` remains unchanged.

```python
comment.replace("Census","Gallup poll")
```

More terminology. The methods `upper()` and `lower()`  don't take any *arguments*, whereas `replace()` takes two - one for the substring we are looking to replace, and one for what to replace it with.

It is important to know that each type of object has different data and different methods associated with it. For example, it makes sense to make `comment`, a string, all caps...

```python
comment.upper()
```

... but it does not make sense to try that with an integer value.

```python
last_3.upper()
```

Your first Python error!

Go to the last line of the error and you will usually be able to read what the problem is. In this case our object `last_3` is of type `int` or integer, and there is no attribute (data or method) associated with the name `upper`.
